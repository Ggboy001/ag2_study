```python
from unstructured.partition.pdf import partition_pdf
from unstructured.staging.base import elements_to_json

file_elements = partition_pdf(
    filename="/content/drive/MyDrive/dataset/finacial_dataset/SEC Filings - AAPL, XOM, JPM/Apple/10K/AAPL_10K_2023.pdf",
    strategy="hi_res",
    languages=["eng"],
    infer_table_structure=True,
    extract_images_in_pdf=True,
    extract_image_block_output_dir="/content/drive/MyDrive/dataset/finacial_dataset/SEC Filings - AAPL, XOM, JPM/Apple/parsed_pdf_info_2023",
    extract_image_block_types=["Image", "Table"],
    extract_forms=False,
    form_extraction_skip_tables=False,
)
```

```python
import json

output_elements = []
keys_to_extract = ["element_id", "text", "type"]
metadata_keys = ["page_number", "parent_id", "image_path"]
text_types = set(["Text", "UncategorizedText", "NarrativeText"])
element_length = len(file_elements)
for idx in range(element_length):
    data = file_elements[idx].to_dict()
    new_data = {key: data[key] for key in keys_to_extract}
    metadata = data["metadata"]
    for key in metadata_keys:
        if key in metadata:
            new_data[key] = metadata[key]
    if data["type"] == "Table":
        if idx > 0:
            pre_data = file_elements[idx - 1].to_dict()
            if pre_data["type"] in text_types:
                new_data["text"] = pre_data["text"] + new_data["text"]
        if idx < element_length - 1:
            post_data = file_elements[idx + 1].to_dict()
            if post_data["type"] in text_types:
                new_data["text"] = new_data["text"] + post_data["text"]
    output_elements.append(new_data)

with open("simple_elements.json", "w", encoding="utf-8") as file:
    json.dump(output_elements, file, indent=4)
```

```python
import json

# 读取 JSON 文件
try:
    with open('/content/drive/MyDrive/dataset/finacial_dataset/SEC Filings - AAPL, XOM, JPM/Apple/json/2024.json', 'r', encoding='utf-8') as f:
        file_elements = json.load(f)
except FileNotFoundError:
    print("错误: 文件未找到!")
except json.JSONDecodeError:
    print("错误: 无法解析 JSON 文件!")
else:
    output_elements = []
    keys_to_extract = ["element_id", "text", "type"]
    metadata_keys = ["page_number", "parent_id", "image_path"]
    text_types = set(["Text", "UncategorizedText", "NarrativeText"])
    element_length = len(file_elements)
    for idx in range(element_length):
        data = file_elements[idx]
        new_data = {key: data[key] for key in keys_to_extract if key in data}
        metadata = data.get("metadata", {})
        for key in metadata_keys:
            if key in metadata:
                new_data[key] = metadata[key]
        if data.get("type") == "Table":
            if idx > 0:
                pre_data = file_elements[idx - 1]
                if pre_data.get("type") in text_types:
                    new_data["text"] = pre_data.get("text", "") + new_data.get("text", "")
            if idx < element_length - 1:
                post_data = file_elements[idx + 1]
                if post_data.get("type") in text_types:
                    new_data["text"] = new_data.get("text", "") + post_data.get("text", "")
        output_elements.append(new_data)

    with open("/content/drive/MyDrive/dataset/finacial_dataset/SEC Filings - AAPL, XOM, JPM/Apple/json/simple_elements.json", "w", encoding="utf-8") as file:
        json.dump(output_elements, file, indent=4)
```
